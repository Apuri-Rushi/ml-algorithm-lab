```python
ranked_models = comparison_df.sort_values(
    by="Test MSE",
    ascending=True
).reset_index(drop=True)

display(ranked_models[
    ["Model", "Test MSE", "Test RMSE", "Test R2"]
].round(5))

print("Best model:", ranked_models.iloc[0]["Model"])
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Test MSE"
)

plt.title("Testing MSE Comparison")
plt.xlabel("Regression Model")
plt.ylabel("Test MSE")

plt.xticks(rotation=30, ha="right")
plt.yscale("log")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="Test R2"
)

plt.title("Testing R² Comparison")
plt.xlabel("Regression Model")
plt.ylabel("Test R²")

plt.xticks(rotation=30, ha="right")
# plt.yscale("log")
plt.tight_layout()
plt.show()
```

```python
x = np.arange(len(comparison_df))
width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x - width/2,
    comparison_df["Train MSE"],
    width,
    label="Training MSE"
)

plt.bar(
    x + width/2,
    comparison_df["Test MSE"],
    width,
    label="Testing MSE"
)

plt.xticks(x, comparison_df["Model"], rotation=30, ha="right")

plt.ylabel("MSE")
plt.title("Training vs Testing Error Comparison")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
comparison_df.to_csv(
    "regression_model_comparison.csv",
    index=False
)

print("Comparison results saved successfully!")
```
